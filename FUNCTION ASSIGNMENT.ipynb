Q.1 What is the difference between a function and a method in Python?

ANS. In Python, a **function** is a block of code defined using `def` that performs a specific task and can be called independently, whereas a **method** is a function that is defined inside a class and is always associated with an object of that class. For example, `def add(a, b): return a+b` is a function that can be called directly, while inside a class `Calculator`, defining `def multiply(self, a, b): return a*b` makes it a method, which must be called using an object like `calc.multiply(3,5)`. Thus, the main difference is that functions are independent, while methods work with objects.


Q.2 . Explain the concept of function arguments and parameters in Python.

ANS. In Python, **parameters** are the names listed in a function’s definition that act as placeholders for the values the function will receive, while **arguments** are the actual values or data you pass to the function when calling it. For example, in `def greet(name):`, `name` is a parameter, and when we call `greet("John")`, the value `"John"` is the argument. Parameters define what kind of input a function can take, and arguments provide the real input during execution. This concept allows functions to be flexible and reusable with different values.


Q.3  What are the different ways to define and call a function in Python?

ANS. In Python, a function can be defined using the `def` keyword or by using the `lambda` keyword for anonymous functions. The `def` keyword is used to create normal functions with a name, parameters, and a block of code, while `lambda` creates small, unnamed functions for short operations. Once defined, a function can be called simply by writing its name followed by parentheses and passing the required arguments if any. For example, `def add(a, b): return a+b` defines a function that can be called as `add(3, 5)`, and a lambda function like `square = lambda x: x*x` can be called as `square(4)`. Thus, functions in Python can be defined either normally with `def` or anonymously with `lambda`, and are always called by their name followed by parentheses with arguments.


Q.4 What is the purpose of the `return` statement in a Python function?

ANS. The **`return` statement** in a Python function is used to send a value or result back to the caller after the function has finished executing. It allows the function to produce an output that can be stored in a variable, used in expressions, or passed to other functions. Without `return`, a function will by default return `None`, meaning it performs an action but does not give back any usable value. For example, `def add(a, b): return a + b` sends the sum back to the caller, so when we write `result = add(3, 5)`, the value `8` is stored in `result`. Thus, `return` makes functions more flexible and reusable by providing outputs.


Q.5  What are iterators in Python and how do they differ from iterables?

ANS. In Python, an **iterable** is any object that can be looped over, such as lists, tuples, strings, or sets, because they implement the `__iter__()` method which returns an iterator. An **iterator**, on the other hand, is an object that represents a stream of data and remembers its state, allowing you to fetch elements one at a time using the `next()` function until all items are exhausted. Iterators implement both `__iter__()` and `__next__()` methods, whereas iterables only provide a way to get an iterator. For example, a list is an iterable, but when we call `iter([1, 2, 3])`, it gives an iterator that can return elements one by one using `next()`. Thus, the key difference is that **iterables are containers you can loop through, while iterators are the objects that actually produce the values during iteration**.


Q.6  Explain the concept of generators in Python and how they are defined.

ANS. In Python, **generators** are a special type of iterable that allow you to generate values on the fly instead of storing them all in memory at once. They are defined using a normal function but with the `yield` keyword instead of `return`. Each time the generator function is called, execution pauses at the `yield` statement, returning the yielded value to the caller, and resumes from that point on the next call. This makes generators very memory-efficient, especially when working with large datasets or infinite sequences, since values are produced one at a time rather than precomputed and stored in a list. Generators can be defined in two ways: by writing a function with `yield`, or by using **generator expressions**, which look similar to list comprehensions but use parentheses instead of square brackets.


Q.7 What are the advantages of using generators over regular functions?

ANS. Generators in Python have several advantages over regular functions. The main benefit is **memory efficiency**, as they yield values one at a time using the `yield` keyword instead of generating and storing the entire sequence in memory, which is useful when working with large datasets or infinite sequences. Generators also provide **lazy evaluation**, meaning values are produced only when required, which improves performance and reduces overhead. They make code **simpler and cleaner** compared to implementing iterators manually with classes. Additionally, generators maintain their **state automatically** between iterations, eliminating the need to track variables explicitly. Overall, generators are a more efficient and elegant way to handle iteration compared to regular functions.


Q.8 What is a lambda function in Python and when is it typically used?

ANS. A lambda function in Python is an anonymous, single-line function defined using the keyword `lambda` instead of the usual `def`. It can take any number of arguments but can only contain one expression, which is implicitly returned. For example, `lambda x, y: x + y` creates a function that adds two numbers. Lambda functions are typically used when a small, temporary function is needed without formally defining it, such as in functional programming with `map()`, `filter()`, and `sorted()` for short inline operations. They are most useful for improving code readability and reducing boilerplate when the function logic is simple and used only once.


Q.9 Explain the purpose and usage of the `map()` function in Python

ANS. The `map()` function in Python is used to apply a given function to each item of an iterable (like a list, tuple, or set) and returns an iterator with the transformed results. Its purpose is to simplify the process of performing the same operation on multiple elements without writing explicit loops. The general syntax is `map(function, iterable)`, where the function is applied to every element of the iterable. For example, `map(lambda x: x**2, [1, 2, 3])` will produce `[1, 4, 9]` when converted to a list. It is typically used for concise, functional-style programming when transformations need to be applied efficiently across collections.


Q.10 What is the difference between `map()`, `reduce()`, and `filter()` functions in Python?

ANS. In Python, `map()`, `reduce()`, and `filter()` are functional programming tools that operate on iterables but serve different purposes. The `map()` function applies a given function to each element of an iterable and returns a new iterator with the transformed values, effectively performing element-wise transformations. The `filter()` function, on the other hand, applies a Boolean function to each element and returns only those elements for which the function evaluates to `True`, thus selectively filtering items. `reduce()`, available in the `functools` module, repeatedly applies a binary function to the elements of an iterable to reduce it to a single cumulative value, such as summing a list of numbers or computing a product. Together, these functions allow concise, functional-style processing of data with transformations, filtering, and aggregation.

Q.1 Write a Python function that takes a list of numbers as input and returns the sum of all even numbers in
the list.

````python
def sum_even_numbers(numbers):
    """
    This function takes a list of numbers and returns the sum of all even numbers.
    """
    total = 0
    for num in numbers:
        if num % 2 == 0:
            total += num
    return total

# Example usage
nums = [10, 15, 22, 33, 40]
print("Sum of even numbers:", sum_even_numbers(nums))
````

````python
def square_list(numbers):
    """
    This function takes a list of integers and returns a new list containing the square of each number.
    """
    squared = [num**2 for num in numbers]
    return squared

# Example usage
nums = [2, 5, 7, 10]
print("Squares of the list:", square_list(nums))
````

Q.4 Write a Python function that checks if a given number is prime or not from 1 to 200.

````python
def is_prime(n):
    """
    This function checks if a given number n is prime.
    Returns True if prime, False otherwise.
    """
    if n <= 1:
        return False
    for i in range(2, int(n**0.5) + 1):
        if n % i == 0:
            return False
    return True

# Example usage: Check all numbers from 1 to 200
for num in range(1, 201):
    if is_prime(num):
        print(num, "is prime")
````

Q.5  Create an iterator class in Python that generates the Fibonacci sequence up to a specified number of
terms.

````python
class FibonacciIterator:
    def __init__(self, n):
        self.n = n        # total number of terms
        self.index = 0    # current index
        self.a = 0        # first Fibonacci number
        self.b = 1        # second Fibonacci number

    def __iter__(self):
        return self

    def __next__(self):
        if self.index >= self.n:
            raise StopIteration
        if self.index == 0:
            self.index += 1
            return self.a
        elif self.index == 1:
            self.index += 1
            return self.b
        else:
            next_value = self.a + self.b
            self.a, self.b = self.b, next_value
            self.index += 1
            return next_value

# Example usage:
n_terms = 10
fib_iter = FibonacciIterator(n_terms)

for num in fib_iter:
    print(num, end=' ')

````

Q.6 Write a generator function in Python that yields the powers of 2 up to a given exponent.

````python
def powers_of_two(max_exponent):
    """
    Generator that yields powers of 2 from 0 to max_exponent (inclusive)
    """
    for i in range(max_exponent + 1):
        yield 2 ** i

# Example usage:
for value in powers_of_two(5):
    print(value, end=' ')
````

Q.7 Implement a generator function that reads a file line by line and yields each line as a string.

````python
def read_file_lines(file_path):
    """
    Generator that yields each line of a file as a string.
    """
    with open(file_path, 'r') as file:
        for line in file:
            yield line.rstrip('\n')  # Remove newline character

# Example usage:
file_path = 'example.txt'  # Replace with your file path

for line in read_file_lines(file_path):
    print(line)
````

Q.9 Write a Python program that uses `map()` to convert a list of temperatures from Celsius to Fahrenheit.

````python
# Function to convert Celsius to Fahrenheit
def celsius_to_fahrenheit(c):
    return (c * 9/5) + 32

# List of temperatures in Celsius
celsius_temps = [0, 20, 37, 100]

# Use map() to convert each temperature to Fahrenheit
fahrenheit_temps = list(map(celsius_to_fahrenheit, celsius_temps))

print(fahrenheit_temps)
````

Q.10 Create a Python program that uses `filter()` to remove all the vowels from a given string.

````python
# Function to check if a character is not a vowel
def is_not_vowel(char):
    vowels = "aeiouAEIOU"
    return char not in vowels

# Input string from the user
input_string = input("Enter a string: ")

# Using filter() to remove vowels
result = ''.join(filter(is_not_vowel, input_string))

# Display the result
print("String after removing vowels:", result)
````
